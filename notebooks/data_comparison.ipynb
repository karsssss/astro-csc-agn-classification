```python
import numpy as np
import pandas as pd
from astropy.io import fits
from astropy.table import Table


path = '../data/raw/Multiwave_UMAP_6_26_26.csv'
path_old1 = '../data/raw/Multiwave_UMAP_6_22_26.csv'
path_old2 = '../data/raw/Multiwave_6_12_26.fits'


# Processing 2 old datasets
df_old1 = pd.read_csv(path_old1)
with fits.open(path_old2) as hdul:
    table = Table(hdul[1].data)
    df_old2 = table.to_pandas()
    
df_old1 = df_old1.loc[:, ~df_old1.columns.str.startswith("Unnamed")]
df_old1 = df_old1.set_index('name')
df_old1 = df_old1.rename({'IR_agn': 'ir_agn'}, axis=1)

df_old2 = df_old2.loc[:, ~df_old2.columns.str.startswith("Unnamed")]
df_old2 = df_old2.set_index('name')
df_old2 = df_old2.rename({'IR_agn': 'ir_agn'}, axis=1)


# Processing the most recent dataset
df = pd.read_csv(path)

df = df.loc[:, ~df.columns.str.startswith("Unnamed")]
df = df.set_index('name')
df = df.rename({'IR_agn': 'ir_agn'}, axis=1)



display(df.shape)
display(df_old1.shape)
display(df_old2.shape)

new_cols01 = [col for col in df.columns if col not in df_old1.columns]
new_cols02 = [col for col in df.columns if col not in df_old2.columns]
new_cols12 = [col for col in df_old1.columns if col not in df_old2.columns]

print(new_cols01)
print(new_cols02)
print(new_cols12)

new_col = ['non_agn', 'zErr_normgal', 'z_code_normgal', 'opt_r', 'r_code_normgal', 'opt_g', 'g_code_normgal', 'CSC_b_lum']
display(df[new_col].descirbe(20))


# Treating ' ' as NaN
object_cols = df.select_dtypes(include='object').columns

df[object_cols] = df[object_cols].astype('string').apply(lambda col: col.str.strip())
df[object_cols] = df[object_cols].replace("", np.nan)



# List of columns by dtype
for i in df.dtypes.unique():
    print(f'Columns with {i} dtype:', df.select_dtypes(include=i).columns.tolist())

# Missing value proportion DataFrame
missing = pd.DataFrame({
    'NaN perc': (df.isna().sum() / df.shape[0] * 100).round(2), 
    'missing cnt': df.isna().sum(), 
    'non-missing cnt': df.notna().sum(),
    }).sort_values('NaN perc', ascending=False)

for i in range(0, 100, 25):
    print(f'Columns with NaN values percentage between {i} and {i + 25}: ', end="") 
    print(missing[(missing['NaN perc'] >= i) & 
          (missing['NaN perc'] < i + 25)].shape[0])

missing
```
